Investment Potfolio Tracker

Problem Statement

It would be difficult to the investors to track & analyse their overall portfolio who invests in mutiple platforms & assets. 

Solution

This solution helps them to track all their investments at one place in one tap & avoid the chaos of calculating seperately by using pen,paper or Excel. It saves the valuable time of the investors & eliminating the chance of wrong & missing some records, A every penny is important, wrong results might effects their financial decisions. 

```python
#Investment Portfolio Tracker
portfolio = []

num_stocks = int(input("How many stocks do you want to add? "))

for i in range(num_stocks):
    print(f"\nStock {i+1}")

    asset = input("Asset Name: ")
    quantity = float(input("Quantity: "))
    buy_price = float(input("Buying Price ($): "))
    current_price = float(input("Current Price ($): "))

    investment = quantity * buy_price
    current_value = quantity * current_price
    profit_loss = current_value - investment

    stock = {"asset": asset,"quantity": quantity,"buy_price": buy_price,"current_price": current_price,"investment": investment,"current_value": current_value,"profit_loss": profit_loss}

    portfolio.append(stock)


for stock in portfolio:
    total_investment += stock["investment"]
    total_value += stock["current_value"]
    total_gain_loss += stock["profit_loss"]

print("\n" + "=" * 60)
print("PORTFOLIO ANALYTICS REPORT")
print("=" * 60)

for stock in portfolio:
  
    if total_value != 0:
        allocation = (stock["current_value"] / total_value) * 100
    else:
        allocation = 0

    if stock["investment"] != 0:
        return_percent = (stock["profit_loss"] / stock["investment"]) * 100
    else:
        return_percent = 0

    print(f"\nAsset: {stock['asset']}")
    print(f"Investment: ${stock['investment']:.2f}")
    print(f"Current Value: ${stock['current_value']:.2f}")
    print(f"Profit/Loss: ${stock['profit_loss']:.2f}")
    print(f"Return %: {return_percent:.2f}%")
    print(f"Holding %: {allocation:.2f}%")

print("\n" + "=" * 60)
print(f"Total Investment: ${total_investment:.2f}")
print(f"Total Portfolio Value: ${total_value:.2f}")
print(f"Total Gain/Loss: ${total_gain_loss:.2f}")

if total_investment != 0:
    overall_return = (total_gain_loss / total_investment) * 100
else:
    overall_return = 0
print(f"Overall Return: {overall_return:.2f}%")
print("=" * 60)
```
